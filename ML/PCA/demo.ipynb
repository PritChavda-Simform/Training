```python
import numpy as np
import matplotlib as mlt
import pandas as pd
import seaborn as sns
%matplotlib inline 
```

```python
from  sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
cancer
```

```python
cancer.keys()
```

```python
print(cancer['DESCR'])
```

```python
df = pd.DataFrame(cancer['data'],columns=cancer['feature_names'])
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
std = StandardScaler()

sklear = StandardScaler()
# df = sklear.fit_transform(df)
sklear.fit(df)
```

```python
scaled_data = sklear.transform(df)
```

```python
scaled_data
```

```python
# df = pd.DataFrame(df , columns=cancer['feature_names'])
```

```python
# df.head()
```

```python
from sklearn.decomposition import PCA
```

```python
pc = PCA(n_components=2)
```

```python
pc.fit(scaled_data)
```

```python
x_pca = pc.transform(scaled_data)
```

```python
x_pca
```

```python
x_pca = pd.DataFrame(x_pca, columns= [0,1])
```

```python
sns.scatterplot(x = x_pca[0],y = x_pca[1] , hue = cancer['target'], data=x_pca)
```
